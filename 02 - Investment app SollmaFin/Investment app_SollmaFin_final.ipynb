#  Исследование результатов А/В-теста и поиск инсайтов

SollmaFin — международное мобильное приложение для онлайн-торговли финансовыми активами. Пользователи могут инвестировать деньги в акции, валюту, криптовалюту, биржевые фонды (ETF) и другие активы. Целевая аудитория приложения — начинающие инвесторы. Получить доступ к финансовым рынкам можно со смартфона либо через веб-версию. 

Приложение ориентировано на рынок Латинской Америки с акцентом на четыре страны: Мексику, Бразилию, Колумбию и Аргентину — там быстро растёт интерес к финтеху и инвестициям, а экономики очень разные. 

Появилась гипотеза о недостаточной финансовой грамотности пользователей: они не понимают, как работать с активами разной степени риска. Клиенты покупают активы с высоким риском и теряют деньги, что снижает вовлечённость.
Команда продукта решила обновить онбординг пользователей и добавить в него детальную информацию о различиях финансовых активов и связанных с ними рисках. Однако появилось опасение, что углублённый онбординг отпугнёт клиентов от пополнения депозита и более рискованных вложений. 

Чтобы разобраться, команда решила провести A/B-эксперимент. 

### А/В-эксперимент

В эксперименте участвовали новые пользователи, которые зарегистрировались в приложении со 2 по 15 июня 2025 года. 
Пользователей случайным образом разделили на две равные группы: 
- Контрольная группа проходила стандартный онбординг без обязательного обучения.
- Тестовая группа проходила обновлённый онбординг с подробной информацией об активах и связанных с ними рисках. 

После разделения активность пользователей анализировали в течение недели.

Цель эксперимента — оценить, как обновлённый онбординг влияет на поведение пользователей и их дальнейшую инвестиционную активность.

При внедрении новой фичи команда продукта выдвинула такие гипотезы:
- Гипотеза роста: обучающий онбординг помогает пользователям лучше понимать принципы инвестирования, поэтому они будут чаще открывать второй депозит.
- Гипотеза риска: информация о возможных потерях и высоких рисках отпугнёт некоторых новичков, особенно самых осторожных, что снизит конверсию в первый депозит.
- Дополнительная гипотеза: после нового онбординга пользователи, которые выбрали высокорискованные активы, будут чаще, чем раньше, возвращаться и открывать второй депозит. При старом онбординге пользователи часто покупали активы с высоким риском без понимания последствий. Это приводило к потерям и оттоку после первого депозита. 

Чтобы увидеть полную картину, недостаточно следить за метриками конверсии. Команда выбрала для эксперимента такой набор метрик: 
- Ключевая метрика — средняя сумма всех депозитов на одного пользователя (включая тех, кто установил приложение или открыл веб-версию).
- Барьерная метрика — конверсия из регистрации в первый депозит.
- Вспомогательная метрика 1 — конверсия из первого депозита во второй.
- Вспомогательная метрика 2 — средняя сумма всех депозитов на пользователя, который открыл хотя бы один депозит.

Ожидалось, что ключевая и барьерная метрики не упадут, а вспомогательные покажут значительный рост.

Стоит изучить источники привлечения.

Посмотрим, сколько из новичков вносят первый и второй депозит в процентном соотношении.

```python
df_hist_install_users = (
    df_hist[df_hist['event_name'] == 'install / open_web']
    .groupby('user_id')['event_ts']
    .min()
    .reset_index()
    .rename(columns = {'event_ts': 'install_ts'})
)

df_hist_deposit_users = (
    df_hist[df_hist['event_name'] == 'first_deposit']
    .groupby('user_id')['event_ts']
    .min()
    .reset_index()
    .rename(columns = {'event_ts': 'deposit_ts'})
)

merged = df_hist_install_users.merge(df_hist_deposit_users, on = 'user_id', how = 'left')

merged['install_date'] = merged['install_ts'].dt.date
merged['deposit_date'] = merged['deposit_ts'].dt.date

df_hist_stats = (
    merged
    .groupby('install_date')
    .agg(
        total_installed = ('user_id', 'count'),
        made_deposit = ('deposit_date', 'count') 
    )
    .reset_index()
)

df_hist_stats['conversion_rate'] = (
    df_hist_stats['made_deposit'] / df_hist_stats['total_installed'] * 100
).round(2)

df_hist_stats = df_hist_stats.sort_values('install_date').reset_index(drop = True)

plt.figure(figsize = (12, 6))

sns.lineplot(
    data = df_hist_stats, 
    x = 'install_date', 
    y = 'conversion_rate',
    label = 'new_per_day', 
    color = 'salmon', 
    marker = 'o',
    linewidth = 2
)

plt.grid(alpha = .4)
plt.xticks(rotation = 45)

plt.xlabel('Даты')
plt.ylabel('Процент')
plt.title('Динамика новых пользователей, которые сделали первый депозит')

plt.tight_layout()
plt.show()
```

```python
df_hist_deposit_users_2 = (
    df_hist[df_hist['event_name'] == 'second_deposit']
    .groupby('user_id')['event_ts']
    .min()
    .reset_index()
    .rename(columns = {'event_ts': 'deposit_ts'})
)

merged = df_hist_install_users.merge(df_hist_deposit_users_2, on = 'user_id', how = 'left')

merged['install_date'] = merged['install_ts'].dt.date
merged['deposit_date'] = merged['deposit_ts'].dt.date

df_hist_stats_2 = (
    merged
    .groupby('install_date')
    .agg(
        total_installed = ('user_id', 'count'),
        made_deposit = ('deposit_date', 'count') 
    )
    .reset_index()
)

df_hist_stats_2['conversion_rate'] = (
    df_hist_stats_2['made_deposit'] / df_hist_stats_2['total_installed'] * 100
).round(2)

df_hist_stats_2 = df_hist_stats_2.sort_values('install_date').reset_index(drop = True)

plt.figure(figsize = (12, 6))

sns.lineplot(
    data = df_hist_stats_2, 
    x = 'install_date', 
    y = 'conversion_rate',
    label = 'new_per_day', 
    color = 'salmon', 
    marker = 'o',
    linewidth = 2
)

plt.grid(alpha = .4)
plt.xticks(rotation = 45)

plt.xlabel('Даты')
plt.ylabel('Процент')
plt.title('Динамика новых пользователей, которые сделали второй депозит')

plt.tight_layout()
plt.show()
```

Из зарегестрировавшихся пользователей доля тех, кто в последующем времени сделал первый депозит в среднем 65-70%, а второй - около 10-15%. При решении построим две воронки: 
- классическую воронку — доля от первого шага в процентах
- step-by-step — с конверсией от предыдущего этапа.

Предполагается, что пользователь проходит такой путь к оформлению доставки блюда из ресторанов:
1. `install / open_web` — запуск приложения/первый заход на сайт (шаг `introduction` отнесём сюда же).
2. `registration` — регистрация.
3. `main_page` — переход на главную страницу.
4. `onboarding_complete` — завершение этапа онбординга.
5. `first_deposit` — внесение первого депозита.
6. `asset_purchase` — покупка актива.
7. `second_deposit` — внесение второго депозита.

**Задача 2.3.** Влияние уровня риска актива на открытие второго депозита.

Изучим, как уровень риска первого актива влияет на конверсию во второй депозит.  

```python
first_deposit = (
    df_hist[df_hist['event_name'] == 'asset_purchase']
    .sort_values('event_ts')
    .groupby('user_id')
    .first()
    [['event_ts', 'risk_level']]
    .reset_index()
)

second_deposit = (
    df_hist[df_hist['event_name'] == 'second_deposit']
    .groupby('user_id')['event_ts']
    .min()
    .reset_index()
)

deposit_merg = first_deposit.merge(second_deposit, on = 'user_id', how = 'left')

deposit = (
    deposit_merg
    .groupby('risk_level')
    .agg(
        first_deposit_users = ('user_id', 'count'),
        second_deposit_users = ('event_ts_y', 'count')
    )
    .reset_index()
)

deposit['rate'] = (deposit['second_deposit_users'] / deposit['first_deposit_users'] * 100).round(1)
deposit = deposit.sort_values('rate', ascending = False).reset_index(drop = True)
deposit
```

Самый популярный первый актив пользователей – с высоким риском, после этого второй раз депозит пополняет только четверть из них.

```python
plt.figure(figsize = (12, 6))

plt.bar(
    deposit['risk_level'], 
    deposit['rate'], 
    color = 'salmon'
)

plt.title('Конверсия во второй депозит в зависимости от риска первого актива', fontsize = 14)
plt.xlabel('Риск первого актива', fontsize = 12)
plt.ylabel('Конверсия (%)', fontsize = 12)

plt.grid(True, linestyle = '--', alpha = 0.5)
plt.xticks(rotation = 45, ha = 'right')

plt.tight_layout()
plt.show()
```

Самый непопулярный тип первого актива обеспечивает самую высокую возвращаемость. После покупки актива с высоким риском второй депозит вносит только четверть пользователей.

**Задача 2.4.** Анализ ключевой метрики на исторических данных.

Рассчитаем значение ключевой метрики — средней суммы всех депозитов на одного пользователя. Учитываем всех пользователей, которые взаимодействовали с сервисом. Это позволит зафиксировать базовый уровень монетизации. 

```python
avg_dep_per_user = df_hist['amount'].sum() / df_hist['user_id'].nunique()
avg_dep_per_user
```

```python
df_hist_avg_dep = df_hist
df_hist_avg_dep['month'] = df_hist['first_ts'].dt.to_period('M')
df_hist_avg_dep
```

```python
month_avg_dep = (
    df_hist_avg_dep
    .groupby('month')
    .agg(
        total_amount = ('amount', 'sum'),
        users = ('user_id', 'nunique'),
        avg_dep_per_user = ('amount', 'mean'),
    )
    .assign(key_metric = lambda x: x['total_amount'] / x['users'])
    .reset_index()
)

month_avg_dep
```

В качестве значения ключевой метрики (средней суммц всех депозитов на одного пользователя, включая тех, кто установил приложение или открыл веб-версию) будем использовать 91.6.

Также на исторических данных изучили поведение пользователей с депозитами и активами, их склонность к высокому риску. Самое проблемное место воронки - покупка первого актива.

```python
df_hist['amount'].describe()
```

```python
median_dep = df_hist['amount'].median()
q25 = df_hist['amount'].quantile(0.25)
q75 = df_hist['amount'].quantile(0.75)
mean_dep = df_hist['amount'].mean()

fig, ax = plt.subplots(figsize = (12, 6))

n, bins, patches = ax.hist(
    df_hist['amount'].dropna(), 
    bins = 50, 
    color = 'salmon', 
    edgecolor = 'black', 
    alpha = 0.7
)

colors = []
for i in range(len(bins) - 1):
    bin_mid = (bins[i] + bins[i+1]) / 2
    if bin_mid < q25:
        colors.append('lightblue')
    elif bin_mid < median_dep:
        colors.append('yellow')
    elif bin_mid < q75:
        colors.append('lightgreen')
    else:
        colors.append('salmon')

for patch, color in zip(patches, colors):
    patch.set_facecolor(color)

ax.axvline(median_dep, color = 'blue', linestyle = '-', linewidth = 2, label = 'Медиана')
ax.axvline(mean_dep, color = 'red', linestyle = '-', linewidth = 2, label = 'Среднее')

ax.set_title('Распределение сумм депозитов')
ax.set_xlabel('Сумма депозита')
ax.set_ylabel('Количество пользователей')

ax.grid(axis = 'y', linestyle = '--', alpha = 0.5)
ax.legend()
plt.show()
```

Медиана меньше среднего, виден правосторонний перекос распределения из-за крупных депозитов, сектор после 75 квантиля самый размазанный. H0 - Новая фича не влияет на конверсию во второй депозит: доля пользователей, совершивших второй депозит, в тестовой группе не отличается от доли в контрольной группе в каждом сегменте по уровню риска актива.

H1 - Новая фича влияет на конверсию во второй депозит: Доля пользователей, совершивших второй депозит, в тестовой группе статистически значимо отличается от доли в контрольной группе.

Будем использовать z-тесты для каждой группы риска.

```python
high_risk_users = df_abt[
    (df_abt['event_name'] == 'asset_purchase') & 
    (df_abt['risk_level'] == 'high')
]['user_id'].unique()

df_hr_users = df_abt[df_abt['user_id'].isin(high_risk_users)].copy()
deposits = df_hr_users[df_hr_users['event_name'].isin(['first_deposit', 'second_deposit'])].copy()
deposits = deposits.sort_values(['user_id', 'event_ts'])
deposits['deposit_num'] = deposits.groupby('user_id').cumcount() + 1

first_deposits = deposits[deposits['deposit_num'] == 1][['user_id', 'group', 'event_ts']].groupby('user_id').agg(
    first_deposit_ts=('event_ts', 'min'),
    group=('group', 'first')
).reset_index()

second_deposits = deposits[deposits['deposit_num'] == 2][['user_id', 'event_ts']].groupby('user_id')['event_ts'].min().reset_index()
second_deposits.rename(columns = {'event_ts': 'second_deposit_ts'}, inplace = True)
cohort = first_deposits.merge(second_deposits, on = 'user_id', how = 'left')

cohort['converted'] = (
    cohort['second_deposit_ts'].notna() & 
    (cohort['second_deposit_ts'] > cohort['first_deposit_ts'])
).astype(int)

cohort['cohort_date'] = cohort['first_deposit_ts'].dt.date
agg = cohort.groupby('group')['converted'].agg(['sum', 'count']).rename(columns={'sum': 'conversions', 'count': 'total_users'})
agg['conversion_rate'] = agg['conversions'] / agg['total_users']
count = agg.loc[['control', 'test'], 'conversions'].values
nobs = agg.loc[['control', 'test'], 'total_users'].values

z_stat, p_val = proportions_ztest(count, nobs, alternative='two-sided')

conv_control = agg.loc['control', 'conversion_rate']
conv_test = agg.loc['test', 'conversion_rate']
absolute_lift = conv_test - conv_control
relative_lift = 100 * (conv_test / conv_control - 1) if conv_control != 0 else 0

results_df_high_risk = pd.DataFrame([{
    'Segment': 'high_risk',
    'Group A': 'control',
    'Group B': 'test',
    'p-value': p_val,
    'Conv A': conv_control,
    'Conv B': conv_test,
    'Absolute Lift': absolute_lift,
    'Relative Lift (%)': relative_lift
}])

results_df_high_risk
```

Цель эксперимента — оценить, как обновлённый онбординг влияет на поведение пользователей и их дальнейшую инвестиционную активность.

При внедрении новой фичи команда продукта выдвинула такие гипотезы:
- Гипотеза роста: обучающий онбординг помогает пользователям лучше понимать принципы инвестирования, поэтому они будут чаще открывать второй депозит.
- Гипотеза риска: информация о возможных потерях и высоких рисках отпугнёт некоторых новичков, особенно самых осторожных, что снизит конверсию в первый депозит.
- Дополнительная гипотеза: после нового онбординга пользователи, которые выбрали высокорискованные активы, будут чаще, чем раньше, возвращаться и открывать второй депозит. При старом онбординге пользователи часто покупали активы с высоким риском без понимания последствий. Это приводило к потерям и оттоку после первого депозита.

Ключевая метрика — средняя сумма всех депозитов на одного пользователя (включая тех, кто установил приложение или открыл веб-версию). Ожидается, что она не упадёт.

```python
df_abt_agg = (
    df_abt
    .groupby(['user_id', 'group'], as_index=False)['amount']
    .sum()
    .rename(columns={'amount': 'total_deposits_per_user'})
)

df_abt_agg = df_abt_agg[df_abt_agg['total_deposits_per_user'] > 0]

df_abt_agg
```

```python
np.random.seed(41)

control = df_abt_agg[df_abt_agg['group'] == 'control']['total_deposits_per_user'].values
test = df_abt_agg[df_abt_agg['group'] == 'test']['total_deposits_per_user'].values

n_iterations = 1000
boot_diffs = []

for i in range(n_iterations):
    boot_control = np.random.choice(control, size = len(control), replace = True)
    boot_test = np.random.choice(test, size = len(test), replace = True)
    control_25p = np.percentile(boot_control, 25)
    test_25p = np.percentile(boot_test, 25)
    boot_diffs.append(test_25p - control_25p)

boot_diffs = np.array(boot_diffs)
diff_ci = np.percentile(boot_diffs, [2.5, 97.5])

print(f"25-й перцентиль контрольной группы: {np.percentile(control, 25):.2f}")
print(f"25-й перцентиль тестовой группы: {np.percentile(test, 25):.2f}")
print(f"95%-й доверительный интервал разности 25-х перцентилей [{diff_ci[0]:.2f}, {diff_ci[1]:.2f}]")
```
